# Entropy-Based Bankruptcy Prediction using Information Gain in Business Analytics


# Objectives


1.   To understand the concept of Entropy in Business Analytics.
2.   To study the meaning and importance of Information Gain.
3.   To calculate entropy of the bankruptcy dataset.
4.   To calculate Information Gain for each feature column against the target column “Bankrupt?”.
5.   To identify the most important features influencing bankruptcy prediction.
6.   To implement the calculations using Python programming.



```
import pandas as pd
import math
from collections import Counter

# Read the bankruptcy dataset
df = pd.read_csv("Bankrupt.csv")

# Display first 5 rows
print(df.head())

# Display dataset shape
print("Dataset Shape:", df.shape)
```

```
# Entropy Function
def entropy(class_labels):
    total = len(class_labels)
    counts = Counter(class_labels)

    return -sum(
        (count / total) * math.log2(count / total)
        for count in counts.values()
    )


# Information Gain Function
def information_gain(dataframe, feature_column, target_column):

    # Total entropy of target column
    total_entropy = entropy(dataframe[target_column])

    total_len = len(dataframe)

    weighted_entropy = 0.0

    # Calculate weighted entropy for each unique value
    for value in dataframe[feature_column].unique():

        subset = dataframe[dataframe[feature_column] == value]

        subset_entropy = entropy(subset[target_column])

        weighted_entropy += (len(subset) / total_len) * subset_entropy

    # Information Gain
    return total_entropy - weighted_entropy


# Specify Target Column (Y)
target_column = "Bankrupt?"


# Calculate overall entropy of target column
overall_entropy = entropy(df[target_column])


# Store results
results = []

# Loop through all columns except target column
for column in df.columns:

    if column != target_column:

        ig = information_gain(df, column, target_column)

        results.append({
            "Feature Column": column,
            "Information Gain": round(ig, 6)
        })


# Convert results into table
results_df = pd.DataFrame(results)

# Sort by highest information gain
results_df = results_df.sort_values(
    by="Information Gain",
    ascending=False
)

# Display Results
print(f"\nOverall Entropy of '{target_column}': {overall_entropy:.6f}\n")

print("Information Gain of Each Column vs Target:\n")

print(results_df.to_string(index=False))
```
